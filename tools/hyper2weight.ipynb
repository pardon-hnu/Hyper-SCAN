```python
hypergraph_list=[
    # '/home/hnu/Disk0/ParDon/hypergraph/vldb23NeighborCore/real/Enron.hyp',
    # '/home/hnu/Disk0/ParDon/hypergraph/vldb23NeighborCore/real/contact-primary-school.hyp',
    # '/home/hnu/Disk0/ParDon/hypergraph/vldb23NeighborCore/real/congress-bills.hyp',
    # '/home/hnu/Disk0/ParDon/hypergraph/real2/NDC-substances.hyp',
    # '/home/hnu/Disk0/ParDon/hypergraph/vldb23NeighborCore/real/DBLP.hyp',
    # '/home/hnu/Disk0/ParDon/hypergraph/vldb23NeighborCore/real/aminer.hyp',
    # '/home/hnu/Disk0/ParDon/hypergraph/real2/tags-ask-ubuntu.hyp',
    # '/home/hnu/Disk0/ParDon/hypergraph/real2/threads-ask-ubuntu.hyp',
    # '/home/hnu/Disk0/ParDon/hypergraph/sythesis/rpa_t130000000.hyp'
    '/home/hnu/Disk0/ParDon/hypergraph/real2/coauth-DBLP.hyp',
    '/home/hnu/Disk0/ParDon/hypergraph/real2/mag10.hyp',
]

dataset_name=[]
for hypergraph in hypergraph_list:
    temp=re.split('/',hypergraph)
    hyp_name=temp[len(temp)-1]
    name=hyp_name.split('.')
    dataset_name.append(name[0])

print(dataset_name)
```

```python
for i in range(len(hypergraph_list)):
# for i in range(1):
    print('==========================================================')
    print('now deal hypergraph: '+dataset_name[i])
    infile=open(hypergraph_list[i],'r')
    hyperedges=[]
    gs=0
    for line in infile:
        temp_string=line.strip('\n')
        temp=re.split(',',temp_string)
        edge=[]
        if len(temp)!=0:
            for node in temp:
                edge.append(int(node))
                if int(node)>gs:
                    gs=int(node)
        hyperedges.append(edge)
    gs+=1
    infile.close()
    print('==========================================================')
    print('construct link index')
    vertex=[]
    for j in range(gs):
        vertex.append([])
    for j in range(len(hyperedges)):   
        for node in hyperedges[j]:
            vertex[node].append(j)

    print('==========================================================')
    print('construct weighted graph')
    edge={}
    for j in range(gs):
        for e in vertex[j]:
            for v in hyperedges[e]:
                if (j,v) in edge.keys() and j<v:
                    edge[(j,v)]+=1
                if (j,v) not in edge.keys() and j<v:
                    edge[(j,v)]=1
    # print(edge)
    print('general graph: '+dataset_name[i]+' has '+str(gs)+' vertex and '+str(len(edge))+' edges(undirected)')
    print('==========================================================')
    print('output weighted edge txt')
    if not os.path.exists(dataset_name[i]):
        os.makedirs(dataset_name[i])
    edge_file=dataset_name[i]+'/edge.txt'
    ofile1=open(edge_file,'w')
    ofile1.write(str(gs))
    ofile1.write(' ')
    ofile1.write(str(len(edge)))
    ofile1.write('\n')
    for key,value in edge.items():
        for s in key:
            ofile1.write(str(s))
            ofile1.write(' ')
        ofile1.write(str(value))
        ofile1.write('\n')
    ofile1.close()
```
